# Bar plot

```python
language = ["C++", "Python", "C#", "Java", "Rust"]
popularity = [56, 89, 67, 60, 34]
plt.bar(language, popularity, color = 'pink')
plt.show()
```

### Histogram

```python
x
```

### Pie plot

```python
language = ["C++", "Python", "C#", "Java", "Rust"]
popularity = [56, 89, 67, 60, 34]
explode = [0, 0, 0, 0, 0.2]
plt.pie(popularity, labels=language, autopct="%.2f%%", explode=explode)
plt.show()
```

### Showing multiple plot 

```python
plt.figure(figsize=(12, 6))

plt.subplot(2,2,1)
language = ["C++", "Python", "C#", "Java", "Rust"]
popularity = [56, 89, 67, 60, 34]
plt.bar(language, popularity, color = 'pink')
plt.title("Languages and their popularity")
plt.xlabel("Language")
plt.ylabel("Popularity Score")

plt.subplot(2,2,2)
language = ["C++", "Python", "C#", "Java", "Rust"]
popularity = [56, 89, 67, 60, 34]
explode = [0, 0, 0, 0, 0.2]
plt.pie(popularity, labels=language, autopct="%.2f%%", explode=explode)

plt.subplot(2,2,3)
x = np.random.normal(50, 2.5, 100)
plt.hist(x)

plt.subplot(2,2,4)
x = np.arange(-np.pi, 4*np.pi, 0.1)
y = np.sin(x)
plt.plot(x, y)
```
